```python
# Employee Management System

class Employee:

    def __init__(self, name, monthly_salary):
        self.name = name
        self.monthly_salary = monthly_salary

    def annual_salary(self):
        return self.monthly_salary * 12


employees = []

n = int(input("Enter number of employees: "))

for i in range(n):
    print("\nEmployee", i + 1)

    name = input("Enter Employee Name: ")
    salary = int(input("Enter Monthly Salary: "))

    emp = Employee(name, salary)
    employees.append(emp)

print("\nEmployee Details")

for emp in employees:
    annual = emp.annual_salary()

    print("Name:", emp.name)
    print("Monthly Salary:", emp.monthly_salary)
    print("Annual Salary:", annual)

    if annual > 800000:
        print("Status: Annual salary is above ₹8,00,000")
    else:
        print("Status: Annual salary is below ₹8,00,000")

    print("---------------------------")
    
    
```

```python
# Student Records using Tuples

students = []

n = int(input("Enter number of students: "))

# Get student details
for i in range(n):
    name = input("Enter student name: ")
    marks = int(input("Enter marks: "))
    students.append((name, marks))

# Find highest scorer
topper = students[0]

for student in students:
    if student[1] > topper[1]:
        topper = student

# Calculate average
total = 0

for student in students:
    total = total + student[1]

average = total / n

# Display results
print("\nHighest Scoring Student:")
print(topper[0], "-", topper[1])

print("\nAverage Marks:", average)

print("\nStudents Scoring Above Average:")

for student in students:
    if student[1] > average:
        print(student[0], "-", student[1])

# Store topper in dictionary
topper_dict = {
    "Name": topper[0],
    "Marks": topper[1]
}

print("\nTopper Dictionary:")
print(topper_dict)
```

```python
# Product Sales using Dictionary

sales = {}

n = int(input("Enter number of products: "))

# Get product details
for i in range(n):
    product = input("Enter product name: ")

    monthly_sales = []

    for j in range(3):
        sale = int(input("Enter sales for month " + str(j + 1) + ": "))
        monthly_sales.append(sale)

    sales[product] = monthly_sales

summary = {}

highest_product = ""
highest_total = 0

# Calculate total and average
for product in sales:

    total = sum(sales[product])
    average = total / len(sales[product])

    summary[product] = {
        "Total": total,
        "Average": average
    }

    if total > highest_total:
        highest_total = total
        highest_product = product

# Display results
print("\nProduct Summary")

for product in summary:
    print(product)
    print("Total Sales:", summary[product]["Total"])
    print("Average Sales:", summary[product]["Average"])
    print()

print("Highest Selling Product:", highest_product)
print("Highest Total Sales:", highest_total)

print("\nFinal Summary Dictionary")
print(summary)
```

```python
# Word Frequency Program

paragraph = input("Enter a paragraph: ")

# Convert paragraph into words
words = paragraph.split()

# Count total words
total_words = len(words)

print("Total Words:", total_words)

# Find unique words
unique_words = []

for word in words:
    if word not in unique_words:
        unique_words.append(word)

print("\nUnique Words:")
print(unique_words)

# Find frequency of each word
frequency = {}

for word in words:
    if word in frequency:
        frequency[word] = frequency[word] + 1
    else:
        frequency[word] = 1

print("\nWord Frequency:")
for word in frequency:
    print(word, ":", frequency[word])

# Find top 3 frequent words
print("\nTop 3 Most Frequent Words:")

for i in range(3):

    max_word = ""
    max_count = 0

    for word in frequency:
        if frequency[word] > max_count:
            max_count = frequency[word]
            max_word = word

    if max_word != "":
        print(max_word, ":", max_count)
        frequency[max_word] = 0
```

```python
# Menu Driven Program

numbers = []

n = int(input("Enter number of elements: "))

for i in range(n):
    num = int(input("Enter number: "))
    numbers.append(num)


# Remove Duplicate
def remove_duplicate():
    new_list = []

    for num in numbers:
        if num not in new_list:
            new_list.append(num)

    print("After Removing Duplicates:", new_list)


# Display Prime Numbers
def prime_numbers():
    print("Prime Numbers:")

    for num in numbers:
        if num > 1:
            prime = True

            for i in range(2, num):
                if num % i == 0:
                    prime = False
                    break

            if prime:
                print(num)


# Display Even Numbers
def even_numbers():
    print("Even Numbers:")

    for num in numbers:
        if num % 2 == 0:
            print(num)


# Display Odd Numbers
def odd_numbers():
    print("Odd Numbers:")

    for num in numbers:
        if num % 2 != 0:
            print(num)


# Largest and Smallest
def largest_smallest():
    print("Largest Number:", max(numbers))
    print("Smallest Number:", min(numbers))


# Main Menu
while True:

    print("\n----- MENU -----")
    print("1. Remove Duplicate Elements")
    print("2. Display Prime Numbers")
    print("3. Display Even Numbers")
    print("4. Display Odd Numbers")
    print("5. Largest and Smallest")
    print("6. Exit")

    choice = int(input("Enter your choice: "))

    if choice == 1:
        remove_duplicate()

    elif choice == 2:
        prime_numbers()

    elif choice == 3:
        even_numbers()

    elif choice == 4:
        odd_numbers()

    elif choice == 5:
        largest_smallest()

    elif choice == 6:
        print("Program Ended")
        break

    else:
        print("Invalid Choice")
```

```python
# Bank Account Management System

class Bank:

    def __init__(self):
        self.balance = 0
        self.history = []

    # Deposit Money
    def deposit(self, amount):
        if amount > 0:
            self.balance = self.balance + amount
            self.history.append("Deposited: " + str(amount))
            print("Amount Deposited Successfully")
        else:
            print("Invalid Deposit Amount")

    # Withdraw Money
    def withdraw(self, amount):
        if amount <= 0:
            print("Invalid Withdrawal Amount")
        elif amount > self.balance:
            print("Insufficient Balance")
        else:
            self.balance = self.balance - amount
            self.history.append("Withdrawn: " + str(amount))
            print("Amount Withdrawn Successfully")

    # Check Balance
    def check_balance(self):
        print("Available Balance:", self.balance)

    # Transaction History
    def transaction_history(self):
        print("\nTransaction History")

        if len(self.history) == 0:
            print("No Transactions")
        else:
            for item in self.history:
                print(item)


# Create Object
account = Bank()

# Menu
while True:

    print("\n------ BANK MENU ------")
    print("1. Deposit")
    print("2. Withdraw")
    print("3. Check Balance")
    print("4. Transaction History")
    print("5. Exit")

    choice = int(input("Enter your choice: "))

    if choice == 1:
        amount = int(input("Enter Deposit Amount: "))
        account.deposit(amount)

    elif choice == 2:
        amount = int(input("Enter Withdrawal Amount: "))
        account.withdraw(amount)

    elif choice == 3:
        account.check_balance()

    elif choice == 4:
        account.transaction_history()

    elif choice == 5:
        print("Thank You")
        break

    else:
        print("Invalid Choice")
```
